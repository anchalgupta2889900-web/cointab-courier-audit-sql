```python
data = pd.merge(orders, sku, on="SKU", how="left")
data["Total Weight (g)"] =(
    data["Order Qty"] * data["Weight (g)"]
)
data["Weight (kg)"] = data["Total Weight (g)"] /1000
order_weight = data.groupby(
    "ExternOrderNo", as_index=False

)["Weight (kg)"].sum()
print(order_weight.head())


```

```python
order_weight["Weight Slab (kg)"] = (
    np.ceil(order_weight["Weight (kg)"] * 2) / 2

)
print(order_weight.head())
```

```python
invoice = pd.merge(
    invoice,
    pincode,
    on=["Warehouse Pincode","Customer Pincode"],
    how="left",
    suffixes=("","_mapped")

)
if "Zone_mapped" in invoice.columns:
  invoice["Zone"] = invoice["Zone_mapped"]
print(invoice[["Warehouse Pincode", "Customer Pincode","Zone"]].head())
```

```python
order_weight["ExternOrderNo"] = order_weight["ExternOrderNo"].astype(str)
invoice["Order ID"] = invoice["Order ID"].astype(str)
final_data = pd.merge(
    invoice,
    order_weight,
    left_on="Order ID",
    right_on="ExternOrderNo",
    how="left"
)
print(final_data.head())
```

```python
def calcuate_charge(zone,weight,charge_type):
  zone = str(zone).lower()
  fixed = rates.loc[0,charge_type + "_" + zone + "_fixed"]
  additional = rates.loc[0,charge_type + "_" + zone + "_additional"]
  extra_slabs = np.maximum(0,np.ceil(weight / 0.5) - 1)
  return fixed + extra_slabs * additional
```

```python
def expected_charge(row):
  if pd.isna(row["Weight Slab (kg)"]) or pd.isna(row["Zone"]):
    weight = row["Weight Slab (kg)"]
    zone = row["Zone"]
    forward = calculate_charge(zone,weight,"fwd")

    if "RTO" in str(row["Type of Shipment"]).upper():
      rto = calcuate_charge(zone,weight,"fwd")
      return forward + rto
    return forward
final_data["Expected Charge"] = final_data.apply(
    expected_charge,axis=1
)
```

```python
final_data["Difference"] = (
    final_data["Billing Amount (Rs.)"]
    - final_data["Expected Charge"]
)
print(final_data[[
    "Order ID",
    "Expected Charge",
    "Difference"
]].head())
```

```python
final_data["Status"] = np.where(
    final_data["Expected Charge"].isnull(),
    "Check Data",
    np.where(
        np.isclose(final_data["Difference"],0),
        "Correctly Charged",
        np.where(
            final_data["Difference"]>0,
            "Overcharged",
            "Undercharged"
        )
    )
)
print(final_data[["Order ID","Difference","Status"]].head())
```

```python
summary = final_data.groupby("Status").agg(
    Order_Count=("Order ID","count"),
    Total_Expected=("Expected Charge","sum"),
    Total_Actual=("Billing Amount (Rs.)","sum"),
    Total_Dofference=("Difference","sum")

).reset_index()
print(summary)
```

**FINAL REPORT**

```python
with pd.ExcelWriter("Courier_Final_Report.xlsx") as writer:
  final_data.to_excel(
      writer, sheet_name="Order Report", index=False

  )
  summary.to_excel(
      writer,sheet_name="Summary",index=False

  )
  print("Report created successfully")
```

**PROJECT SUMMARY**

**Business Problem:** The main objective of this project is to verify whether the courier company has charged the correct delivery fees for shipment or if they have charged.

**Data Collection:** It utilizes 5 different data sources: Website Order Report,SKU master for product weights,pincode mapping for delivery zones,courier invoice for actual billing details,and the rate card for standard pricing.

**Weight&Slab logic:** The total weight for each order is caluclated by multiplying the product weight by its quantity.This total weight is then rounded up tp the nearest 0.5KG slab,as courier companies base thier billing on weight slabs.

**Zone Identification & Expected Charge:** The delivery zone is identify using the warehoue and customer pincodes via the pincode mapping. Excat expected charges are then calculated by applying the rate card values based on the zone,weight slab,and shipment type.

**Final Output:**By comparing the expected charges wiht the actual charges from the courier invoice, a final order-level audit report and a summary table are genrated to identify any financial discrepancies,overcharge,or undercharges.
